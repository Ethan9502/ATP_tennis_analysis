```python
import pandas as pd

player_info = pd.read_csv("espn_atp_rankings.csv")
tennis_df = pd.read_csv("/Users/chiangethan/.cache/kagglehub/datasets/dissfya/atp-tennis-2000-2023daily-pull/versions/911/atp_tennis.csv")

special_name_cases = {"Etcheverry": "Etcheverry T.", "Struff": "Struff J.L.", "Cerundolo": "Cerundolo J.M.", "Portero": "Martinez P.",
                      "Tirante": "Tirante T.A.", "Burruchaga": "Burruchaga R.", "Yunchaokete": "Bu Y.", "Tseng": "Tseng C.H.", "Vera": "Barrios Vera M.T.",
                      "Wong": "Wong C.", "Galan": "Galan D.E."}

def formatting(name):
    first_last = name.split(" ")
    if first_last[-1] in special_name_cases.keys():
        return special_name_cases[first_last[-1]]
    elif len(first_last) == 3:
        return first_last[1] + " " + first_last[2] + " " + first_last[0][0] + "."
    elif len(first_last) == 4:
        return first_last[1] + " " + first_last[2] + " " + first_last[3] + " " + first_last[0][0] + "."
    return first_last[1] + " " + first_last[0][0] + "."

player_info["Player"] = player_info["Player"].apply(formatting)
```

```python
player_1_list = tennis_df["Player_1"].unique().tolist()
player_2_list = tennis_df["Player_2"].unique().tolist()
all_players = set(player_1_list + player_2_list)

is_in_record = 0
not_in_record = 0
top_150_player = player_info["Player"].unique().tolist()
for p in top_150_player:
    if p in all_players:
        is_in_record += 1
    else:
        print(p)
        not_in_record += 1

print(f"in record: {is_in_record}, not in record: {not_in_record}")
```
